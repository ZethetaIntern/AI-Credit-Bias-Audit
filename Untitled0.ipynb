```python
import pandas as pd
import numpy as np

# 1. Dummy Data Generation (50,000 Data)
np.random.seed(42)
n_samples = 50000

gender = np.random.choice(['Male', 'Female'], size=n_samples, p=[0.55, 0.45])
city_tier = np.random.choice(['Tier 1', 'Tier 2', 'Tier 3'], size=n_samples, p=[0.4, 0.35, 0.25])
age = np.random.randint(21, 65, size=n_samples)
income = np.random.normal(50000, 15000, size=n_samples).clip(15000, 200000)
credit_score = np.random.randint(300, 850, size=n_samples)

# Bias Logic
approval_score = (
    (credit_score / 850) * 0.4 +
    (income / 200000) * 0.3 +
    (gender == 'Male') * 0.15 +
    (city_tier == 'Tier 1') * 0.15
)

loan_approved = (approval_score > 0.55).astype(int)

df = pd.DataFrame({
    'gender': gender,
    'city_tier': city_tier,
    'age': age,
    'annual_income': np.round(income, 2),
    'credit_score': credit_score,
    'loan_approved': loan_approved
})

# CSV Save karne
df.to_csv('loan_data_50k.csv', index=False)
print("✅ Data set tayar zala!")

# 2. Bias Analysis Check
gender_approval = df.groupby('gender')['loan_approved'].mean()
print("\n--- Approval Rates ---")
print(gender_approval)

dpr_gender = gender_approval['Female'] / gender_approval['Male']
print(f"\nDemographic Parity Ratio (Female/Male): {dpr_gender:.2f}")

if dpr_gender < 0.80:
    print("🚨 WARNING: Model madhe Gender Bias (Bhedbhav) ahe!")
else:
    print("✅ SUCCESS: Model fair ahe.")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
import shap
import matplotlib.pyplot as plt

# 1. Data tayar karne
df = pd.read_csv('loan_data_50k.csv')

# Categorical data la numbers madhe convert karne (One-Hot Encoding)
X = pd.get_dummies(df.drop('loan_approved', axis=1), drop_first=True)
y = df['loan_approved']

# 2. XGBoost AI Model Train karne
model = xgb.XGBClassifier(random_state=42)
model.fit(X, y)

print("✅ Model Training Complete!")

# 3. SHAP Explainability Graph Banavne
explainer = shap.TreeExplainer(model)
shap_values = explainer(X)

# Feature Importance Graph (Kontha feature decision sathi saglyat mahatvacha ahe)
plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X, plot_type="bar", show=False)
plt.title("Feature Importance - AI Decision Factors", fontsize=14)
plt.tight_layout()
plt.show()

# SHAP Summary Plot (Bhedbhav kasa hotoay te dakhvnaare graph)
plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X, show=False)
plt.title("SHAP Impact on Loan Approval", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score

# 1. Prediction probabilities milavne
y_prob = model.predict_proba(X)[:, 1]

# Unadjusted (Original) Decisions
df['orig_pred'] = (y_prob > 0.5).astype(int)

# 2. Threshold Adjustment Logic
# Males sathi threshold 0.55 ani Females sathi 0.35 karu (Mhanje Female applicants la fair chance miltil)
male_mask = (df['gender'] == 'Male')
female_mask = (df['gender'] == 'Female')

df['mitigated_pred'] = 0
df.loc[male_mask, 'mitigated_pred'] = (y_prob[male_mask] > 0.55).astype(int)
df.loc[female_mask, 'mitigated_pred'] = (y_prob[female_mask] > 0.35).astype(int)

# 3. Fair Approval Rates Check Karne
orig_female_rate = df[female_mask]['orig_pred'].mean()
orig_male_rate = df[male_mask]['orig_pred'].mean()

mitigated_female_rate = df[female_mask]['mitigated_pred'].mean()
mitigated_male_rate = df[male_mask]['mitigated_pred'].mean()

print("--- BEFORE MITIGATION (Aadhi cha Bias) ---")
print(f"Female Approval Rate: {orig_female_rate:.2%}")
print(f"Male Approval Rate: {orig_male_rate:.2%}")
print(f"DPR: {orig_female_rate / orig_male_rate:.2f}")

print("\n--- AFTER MITIGATION (Bias Kami Kelyanantar) ---")
print(f"Female Approval Rate: {mitigated_female_rate:.2%}")
print(f"Male Approval Rate: {mitigated_male_rate:.2%}")
new_dpr = mitigated_female_rate / mitigated_male_rate
print(f"New Demographic Parity Ratio (DPR): {new_dpr:.2f}")

if new_dpr >= 0.80:
    print("\n🎉 SUCCESS: Model madhil Bias yashasvipane kami zala ahe! (Passed Four-Fifths Rule)")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb

# Data load kara
df = pd.read_csv('loan_data_50k.csv')

# 1. Sample Weights Banavne (Fairness Sathi Data Balance Karne)
# Female approving instances la extra importance/weight dene
df['sample_weight'] = 1.0

# Gender & Approved cases cha balance karne
female_approved = (df['gender'] == 'Female') & (df['loan_approved'] == 1)
df.loc[female_approved, 'sample_weight'] = 4.5  # Female approval weight boost

# Features & Target
X = pd.get_dummies(df.drop(['loan_approved', 'sample_weight'], axis=1), drop_first=True)
y = df['loan_approved']

# 2. Re-weighted Model Train Karne
fair_model = xgb.XGBClassifier(random_state=42)
fair_model.fit(X, y, sample_weight=df['sample_weight'])

# 3. New Fair Predictions Check
df['fair_pred'] = fair_model.predict(X)

fair_female_rate = df[df['gender'] == 'Female']['fair_pred'].mean()
fair_male_rate = df[df['gender'] == 'Male']['fair_pred'].mean()
final_dpr = fair_female_rate / fair_male_rate

print("--- FAIR MODEL MITIGATION RESULTS ---")
print(f"New Female Approval Rate: {fair_female_rate:.2%}")
print(f"New Male Approval Rate: {fair_male_rate:.2%}")
print(f"Final Demographic Parity Ratio (DPR): {final_dpr:.2f}")

if final_dpr >= 0.80:
    print("\n🎉 SUCCESS: Model madhil Bias yashasvipane kami zala ahe! (Passed Four-Fifths Rule)")
else:
    print("\n⚠️ DPR ajunhi target javal aalela nahi ahe.")
```
